## 1. Compiled vs. interpreted — seeing it for yourself

With a compiled language like C, the usual workflow is:

1. Write the code.
2. **Compile** it — a program called a compiler translates your entire code into an executable file (a `.exe` on Windows).
3. **Run** the executable.

Once compiled, the executable is just machine instructions — you can't casually peek inside it while it runs. If you want to know what a variable's value is at some point, you generally need to add a `print` somewhere in the code, then recompile and rerun. Or use a specialized tool called a *debugger*.

Python works differently. There's no separate compilation step you need to run — the Python **interpreter** reads and executes your code directly, one instruction at a time. In a Jupyter notebook like this one, each cell you run is handed to the interpreter immediately, and any variable you create stays alive in memory afterwards. That means you can inspect it at any time, just by asking for it in a new cell — no recompiling, no debugger required.

## 2. Variables — storing information

A **variable** is just a name that points to a piece of information (a *value*) stored in memory. You create one with `=`:

```python
age = 25
```

Python figures out the *type* of a value automatically — you don't have to declare it up front like you would in C (`int age = 25;`). The most common types you'll use are:

| Type | Example | Meaning |
|---|---|---|
| `int` | `25` | a whole number |
| `float` | `3.14` | a decimal number |
| `str` | `"hello"` | text (a *string*) |
| `bool` | `True` / `False` | a yes/no value |
| `list` | `[1, 2, 3]` | an ordered collection of values |

You can check any variable's type with the built-in `type()` function.

````python
age = 25
body_temperature = 201.25
first_name = "Loïc"
is_chomeur = True
scores = [12, 15, 9]

print(type(age))
print(type(body_temperature))
print(type(first_name))
print(type(is_chomeur))
print(type(scores))
````

## 3. Conditionals — making decisions

Programs often need to react differently depending on the situation. That's what `if`, `elif` (short for "else if"), and `else` are for: they let a program run different code depending on whether a condition is `True` or `False`.

```python
if condition:
    ...   # runs only if condition is True
elif other_condition:
    ...   # runs only if condition was False and other_condition is True
else:
    ...   # runs if none of the above were True
```

Conditions are usually built with comparison operators: `==` (equal to), `!=` (different from), `<`, `>`, `<=`, `>=`. Note that the indentation (the spaces before each line) is how Python knows which lines belong to which block — it isn't just style, it's required.

````python
day = "Saturday"

if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
else:
    print("It's a weekday.")
````

````python
mood = "tired"

if mood == "happy":
    message = "This won't last"
elif mood == "tired":
    message = "Fuck your depression"
else:
    message = "I don't give a damn brooo"

print(message)
````

## 4. Lists — storing many things at once

You've already used lists a few times, like `["Emma", "Noah", "Lina"]`. A **list** is an ordered collection of values, written between square brackets `[...]` and separated by commas.

Unlike some languages, a Python list doesn't have to contain only one type of value — you can freely mix strings, numbers, booleans, even other lists, all in the same list:

````python
mixed = ["Loïc", 14, True]

print(type(mixed[0]))
print(type(mixed[1]))
print(type(mixed[2]))
````

Each element has a position, called its **index**, starting at `0`. Use `list[index]` to fetch a specific element — and `list[-1]` to grab the last one, without needing to know how long the list is.

````python
friends = ["Emma", "Noah", "Lina"]

print(friends[0])
print(friends[-1])
print(len(friends))
````

Lists can also grow and shrink after they're created:

- `.append(value)` adds `value` at the end.
- `.remove(value)` removes the first element equal to `value`.
- `.pop(index)` removes *and returns* the element at `index` (the last one, by default).

````python
friends.append("Zoe")
friends
````

````python
friends.remove("Noah")
friends
````

````python
last_out = friends.pop()
last_out
````

## 6. Loops — repeating instructions

Loops let you repeat instructions without copying and pasting them. Python has two kinds:

- **`for`** loops repeat once for each item in a collection (like a list), or a fixed number of times using `range(n)`.
- **`while`** loops repeat for as long as a condition stays `True` — handy when you don't know in advance how many repetitions you'll need.

````python
friends = ["Emma", "Noah", "Lina"]

for friend in friends:
    print(f"Hello, {friend}!")
````

````python
for i in range(3):
    print(f"Repetition number {i}")
````

Note that `range(3)` counts `0, 1, 2` — three values, starting at zero. That trips everyone up at first.

````python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown = countdown - 1

print("Go!")
````

## 7. Functions — packaging instructions

A **function** is a named, reusable block of instructions. You define it with `def`, give it a name and optional **parameters** (inputs), and optionally give back a result with `return`.

````python
def add(a, b):
    result = a + b
    return result

add(3, 5)
````

Calling `add(3, 5)` runs the function's instructions with `a = 3` and `b = 5`, and gives back the value of `result`. Note that `result` only exists *while the function runs* — more on that in a later section.

Functions aren't limited to returning numbers — they can return **any** type of value: a string, a boolean, a list, anything. Here's one that returns a string, and reuses what you just learned about conditionals:

````python
def day_type(day):
    weekend_days = ["Saturday", "Sunday"]
    if day in weekend_days:
        return "It's the weekend!"
    else:
        return "It's a weekday."

day_type("Sunday")
````

The `add()` function above returned an `int`; `day_type()` returns a `str`. The `in` keyword checks whether a value is present in a list — handy for exactly this kind of check.

## 8. Converting between types

Sometimes you have a value of one type but need another — text that represents a number, for instance. Python won't guess for you; you have to convert it explicitly, using `str()`, `int()`, or `float()`.

Let's see what happens if you don't:

````python
age = 25
# This will raise an error: Python won't automatically join text and a number
"I am " + age + " years old"
````

That's a `TypeError`: `+` between a `str` and an `int` doesn't make sense to Python. Convert the number to text first, with `str()`:

````python
age = 25
"I am " + str(age) + " years old"
````

It works the other way too: `int()` and `float()` turn text into numbers, so you can do arithmetic with it:

````python
score_text = "42"
score_number = int(score_text)
score_number + 8
````

````python
float("3.14")
````

## 9. Local vs. global variables — scope

Where a variable is created determines where it can be used — this is called its **scope**.

- A variable created **inside** a function is **local**: it only exists while that function is running, and disappears afterwards. Code outside the function can't see it.
- A variable created **outside** any function is **global**: it can be *read* from inside a function, but a function can't change it unless you explicitly use the `global` keyword (a detail you won't need often as a beginner, but good to know it exists).

Let's see this with an example.

````python
def compute_area(width, height):
    area = width * height  # `area` is local to this function
    return area

compute_area(3, 4)
````

````python
# This will raise an error: `area` doesn't exist outside the function
area
````

That `NameError` is expected — it's exactly the point. `area` was created *inside* `compute_area`, so it only ever existed during that function's execution.

Now compare with a **global** variable:

````python
unit = "meters"

def describe_area(width, height):
    a = width * height
    return f"{a} square {unit}"  # `unit` is read from the global scope

describe_area(3, 4)
````

4. Create a variable `remaining = len(friends)`, then write a `while` loop that prints `"Invitations left: " + str(remaining)` and decreases `remaining` by 1 each time, until it reaches `0`.
